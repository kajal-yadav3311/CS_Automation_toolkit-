```python
import re

def password_strength():

    password = input("Enter your password: ")

    score = 0

    # Check minimum 8 characters
    if len(password) >= 8:
        score += 1

    # Check uppercase letter
    if re.search(r"[A-Z]", password):
        score += 1

    # Check lowercase letter
    if re.search(r"[a-z]", password):
        score += 1

    # Check number
    if re.search(r"[0-9]", password):
        score += 1

    # Check special character
    if re.search(r"[@#$%^&+=!]", password):
        score += 1

    # Display password strength
    if score <= 2:
        print("Password Strength: Weak")
    elif score <= 4:
        print("Password Strength: Moderate")
    else:
        print("Password Strength: Strong")

    print("Requirements satisfied:", score, "/ 5")


# Main program
password_strength()
```

```python

import os

def log_file_scanner():

    print("\n--- LOG FILE SCANNER ---")

    file_name = "security_log.txt"

    if not os.path.isfile(file_name):
        print("Log file not found!")
        return

    keywords = ["ERROR", "FAILED LOGIN", "WARNING"]

    counts = {
        "ERROR": 0,
        "FAILED LOGIN": 0,
        "WARNING": 0
    }

    # Read the log file
    with open(file_name, "r") as file:

        for line in file:
            for keyword in keywords:
                counts[keyword] += line.upper().count(keyword)

    # Display scanning results
    print("\n--- LOG SCANNING RESULTS ---")

    for keyword, count in counts.items():
        print(keyword, ":", count)

    print("\nTotal Alerts:", sum(counts.values()))


if __name__ == "__main__":
    log_file_scanner()
```

```python

import socket
import ipaddress

def vulnerability_scanner():

    print("\n--- BASIC VULNERABILITY SCANNER ---")

    target = input("Enter target IP address: ")

    # Validate IP address
    try:
        ipaddress.IPv4Address(target)
    except ValueError:
        print("Invalid IP address!")
        return

    # Port and service details
    ports = {
        21: "FTP",
        22: "SSH",
        80: "HTTP",
        443: "HTTPS"
    }

    print("\n--- PORT SCANNING RESULTS ---")
    print("Target IP:", target)

    for port, service in ports.items():

        sock = socket.socket(
            socket.AF_INET,
            socket.SOCK_STREAM
        )

        sock.settimeout(1)

        result = sock.connect_ex((target, port))

        if result == 0:
            print(f"Port {port} ({service}): Open")
        else:
            print(f"Port {port} ({service}): Closed")

        sock.close()


if __name__ == "__main__":
    vulnerability_scanner()
```
