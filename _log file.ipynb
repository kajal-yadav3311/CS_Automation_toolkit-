```python

import os

def log_file_scanner():

    print("\n--- LOG FILE SCANNER ---")

    file_name = "security_log.txt"

    if not os.path.isfile(file_name):
        print("Log file not found!")
        return

    keywords = ["ERROR", "FAILED LOGIN", "WARNING"]

    counts = {
        "ERROR": 0,
        "FAILED LOGIN": 0,
        "WARNING": 0
    }

    # Read the log file
    with open(file_name, "r") as file:

        for line in file:
            for keyword in keywords:
                counts[keyword] += line.upper().count(keyword)

    # Display scanning results
    print("\n--- LOG SCANNING RESULTS ---")

    for keyword, count in counts.items():
        print(keyword, ":", count)

    print("\nTotal Alerts:", sum(counts.values()))


if __name__ == "__main__":
    log_file_scanner()
```
